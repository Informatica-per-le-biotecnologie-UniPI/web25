# Eccezioni

Quando riscontriamo un errore, Python ci da dei messaggi di errore, e.g.,

````python
l = list(range(10))
print(l[13])
````

Questi errori sono degli *oggetti* veri e propri: hanno un tipo (`Exception`)e possono essere gestiti:
- possiamo "catturare" un'eccezione, e scrivere del codice per gestire l'errore
- possiamo "lanciare" un'eccezione, indicando il motivo che ha portato all'errore

Per *catturare* un'eccezione, usiamo il costrutto `try/except`:

```python
try:
    l = list(range(10))
    print(l[13])
except IndexError:
    print("Trovato un errore!")
```

Per *lanciare* un'eccezione, usiamo il comando `raise`:

```python
index = 20
l = list(range(10))

if index > len(l) - 1:
    raise IndexError(f"Accesso in posizione {index}, ma solo {len(l) - 1} posizioni disponibili")
else:
    print(l[index])
```

````python
class Penguin:
    def __init__(self, name: str, stomach_capacity: float = 25.0):
        self.name = name
        self.stomach_fill = 0.0
        self.max_stomach_capacity = stomach_capacity
        self.digestive_capacity = 5.0

    def eat_fish(self, fish: float) -> float:
        """Il pinguino mangia il cibo dato, restituisce il grado di sazieta' del pinguino."""
        if self.stomach_fill + fish > self.max_stomach_capacity:
            self.stomach_fill = self.max_stomach_capacity
            raise ValueError(f"{self.name} ha mangiato troppo *burp*")
        else:
            self.stomach_fill += fish

    def fly(self):
        raise ValueError(f"{self.name} non puo' imparare la MN02 VOLO!")

    def digest(self):
        self.stomach_fill = max(0, self.stomach_fill - self.digestive_capacity)
        

carlo = Penguin(name="Carlo", stomach_capacity=15.0)
print(f"Carlo ha mangiato {carlo.stomach_fill} pesci. Puo' ancora mangiare {carlo.max_stomach_capacity - carlo.stomach_fill} pesci")
carlo.eat_fish(3.0)
carlo.eat_fish(5.0)
carlo.eat_fish(12.0)

carlo.digest()
carlo.digest()
carlo.digest()
carlo.eat_fish(9.0)
````

````python
try:
    carlo.eat_fish(5.0)
except ValueError:
    while carlo.stomach_fill > 0:
        carlo.digest()
    
    print(f"Ora lo stomaco e' vuoto: {carlo.stomach_fill}")
````

````python
try:
    carlo.fly()
except ValueError:
    print(f"Carlo non puo' volare :(")
````

# Modules

I moduli incapsulano il codice, e lo rendono facilmente fruibile a terzi. Per installare un modulo, possiamo usare

```shell
pip install {module}
```
, e.g.,
```shell
pip install numpy
```
o
```shell
pip install faker-biology
```
Questo comando installa il modulo, e ora possiamo "importarlo" direttamente nel nostro codice.
```python
import numpy
```

Volendo, possimo anche importare direttamente solo alcune classi e/o funzioni.
```python
# importa solo la funzione default_rng, che crea un generatore randomico
from numpy.random import default_rng


default_rng()
samples = rng.normal(size=2500)  # prendi 2500 valori randomici da una distribuzione normale
samples
```


Una lista di moduli e' disponibile su [pypi](https://pypi.org), da cui potete cercare.
