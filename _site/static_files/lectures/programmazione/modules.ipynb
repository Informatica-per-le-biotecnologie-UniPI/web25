# Eccezioni

Le eccezioni forniscono uno strumento per gestire errori e comportamenti inattesi, e.g.,

- in sintesi proteica, un dato codone non corrisponde a nessun amminoacido
- nella generazione di DNA, inseriamo una base `Z`
- nella ricerca del massimo di un insieme, ci viene dato un insieme vuoto

Ci permettono di indicare casi inattesi e/o particolari, e di gestire gli errori che provocano.
Sono oggetti come tutti gli altri, che vengono creati ad-hoc con un messaggio che indica il perche' del fallimento.

````python
class Seagull:
    def __init__(self, name: str):
        self.name = name
        self.position = "ground"
    
    def fly(self):
        print(f"{self.name}: I believe I can fly...")
        self.position = "fly"
        print(f"{self.name}: I believe I can touch the sky!")

    def land(self):
        self.position = "ground"
    

class Pidgeon:
    def __init__(self, name: str):
        self.name = name
        self.position = "ground"
    
    def fly(self):
        print(f"{self.name}: I believe I can fly...")
        self.position = "fly"
        print(f"{self.name}: I believe I can touch the sky!")

    def land(self):
        self.position = "ground"


class Penguin:
    def __init__(self, name: str):
        self.name = name
        self.position = "ground"
    
    def fly(self):
        print(f"{self.name}: I believe I can fly...")
        raise ValueError(f"Tentativo di volo, ma sono un pinguino :(")

    def land(self):
        self.position = "ground"
````

````python
happy_feet = Penguin("Happy Feet")
carlo = Pidgeon("Carlo")

carlo.fly()
carlo.land()
happy_feet.fly()
happy_feet.land()
````

La creazione dell'eccezione avviene creando un oggetto di tipo `Exception`, e "lanciando" l'eccezione con il comando `raise`.

````python
class Penguin:
    def __init__(self, name: str):
        self.name = name
        self.position = "ground"
    
    def fly(self):
        # creo un'eccezione ad-hoc
        raise NotImplementedError(f"Tentativo di volo, ma sono un pinguino :(")

    def land(self):
        self.position = "ground"

happy_feet = Penguin("Happy Feet")
happy_feet.fly()
````

Alcune classi di eccezioni note:

- `ValueError` da usare quando un valore e' del tipo giusto, ma ha valore inatteso, e.g., un'altezza negativa
- `NotImplementedError` da usare quando avete fornito un metodo, ma non e' ancora implementato
- `KeyError` usato per indicare che l'accesso a un dizionario (o simili) usa una chiave non esistente/non valida
- `IndexError` come `KeyError`, ma per collezioni accessibili con indici numerici, e.g., liste
- `TypeError` usato quando viene trovato un tipo sbagliato, e.g., calcolo della media su stringhe

Esempio.
```python
def average(elements: list) -> float:
    """Calcola il valore medio della lista data.
        :param elements: Lista di cui calcolare la media.
    """
    for e in elements:
        if not isinstance(e, float):
            raise TypeError(f"Valore non numerico: {e}")

    return sum(elements) / len(elements)    
```

## Throw and catch

Le eccezioni risalgono lungo lo stack: se un record di attivazione genera una eccezione e non la gestisce, il record viene rimosso, l'eccezione passata allo stack precedente, e via dicendo.
Se neanche il primo record gestisce l'eccezione, il programma crasha.

Come "gestire" le eccezioni lanciate?
Con il costrutto `try/except/finally`, la prima istruzione che lancia un'eccezione ci rimanda al blocco `except`, dove l'eccezione viene "catturata" e gestita.
Al termine del blocco `try`, o di quello `except`, viene eseguito (l'opzionale) blocco `finally`.

```python
try:
    # il nostro codice
    try_body
except:
    # codice che gestisce l'eccezione
    except_body
finally:
    # codice eseguito al termine del try o dell'except
    finally_body
```

````python
l = list(range(10))
try:
    first_element = l[11]
except:
    print(f"Found exception!")
````

Possiamo anche specificare che eccezione gestire!

````python
l = list(range(10))
try:
    first_element = l[8]
    first_element = l["aa"]
except IndexError as e:
    print(f"Found IndexError exception! {e}")
except TypeError as e:
    print(f"Found TypeError exception! {e}")
````

# Modules

I moduli incapsulano il codice, e lo rendono facilmente fruibile a terzi. Per installare un modulo, possiamo usare

```shell
pip install {module}
```
, e.g.,
```shell
pip install numpy
```
o
```shell
pip install faker-biology
```
Questo comando installa il modulo, e ora possiamo "importarlo" direttamente nel nostro codice.
```python
import numpy
```

Volendo, possimo anche importare direttamente solo alcune classi e/o funzioni.
```python
# importa solo la funzione default_rng, che crea un generatore randomico
from numpy.random import default_rng


default_rng()
samples = rng.normal(size=2500)  # prendi 2500 valori randomici da una distribuzione normale
samples
```


Una lista di moduli e' disponibile su [pypi](https://pypi.org), da cui potete cercare.
